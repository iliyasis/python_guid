# جلسه ۱۰ — لیست‌ها (Lists) در پایتون — بخش ۲

##### در این جلسه یاد می‌گیریم چطور محتوای یک لیست را **تغییر، اضافه، حذف** کنیم، روی آن **حلقه** بزنیم و آن را **مرتب (Sort)** کنیم.

#### موضوعات این جلسه:

- تغییر آیتم‌های لیست
- افزودن آیتم (append, insert, extend)
- حذف آیتم (remove, pop, del, clear)
- حلقه زدن روی لیست
- مرتب‌سازی لیست (sort, reverse)
- تمرین عملی

</div>


<div dir="rtl">

## تغییر آیتم‌های لیست

##### برای تغییر یک آیتم خاص، کافیست به **ایندکس** آن مراجعه کرده و مقدار جدید را جایگزین کنیم.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list[1] = "blackcurrant"
print(my_list)
```

<div dir="rtl">

##### همچنین می‌توان با مشخص کردن یک **بازه‌ی ایندکس**، چند آیتم را همزمان تغییر داد.

</div>


```python
my_list = ["apple", "banana", "cherry", "orange", "kiwi"]
my_list[1:3] = ["blackcurrant", "watermelon"]
print(my_list)
```

<div dir="rtl">
<hr>

## افزودن آیتم — `()append`

##### متد `()append` یک آیتم جدید را به **انتهای** لیست اضافه می‌کند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list.append("orange")
print(my_list)
```

<div dir="rtl">

## افزودن در جایگاه دلخواه — `()insert`

##### متد `()insert` یک آیتم را در **جایگاه مشخص‌شده** اضافه می‌کند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list.insert(1, "orange")
print(my_list)
```

<div dir="rtl">

## افزودن چند آیتم — `()extend`

##### برای افزودن آیتم‌های یک **لیست دیگر** به لیست فعلی، از متد `()extend` استفاده می‌کنیم.

</div>


```python
my_list = ["apple", "banana", "cherry"]
list2 = ["mango", "pineapple", "papaya"]

my_list.extend(list2)
print(my_list)
```

<div dir="rtl">
<hr>

## حذف آیتم — `()remove`

##### متد `()remove` یک آیتم را بر اساس **مقدار** آن حذف می‌کند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list.remove("banana")
print(my_list)
```

<div dir="rtl">

## حذف با ایندکس — `()pop`

##### متد `()pop` آیتم را بر اساس **ایندکس** حذف می‌کند. اگر ایندکسی مشخص نشود، **آخرین آیتم** حذف می‌شود.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list.pop(1)
print(my_list)

#my_list.pop() 
#print(my_list)
```

<div dir="rtl">

## حذف با `del`

##### کلمه‌کلیدی `del` هم می‌تواند یک آیتم مشخص را حذف کند، و هم می‌تواند **کل لیست** را از حافظه پاک کند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
del my_list[0]
print(my_list)
```

<div dir="rtl">

## خالی کردن لیست — `()clear`

##### متد `()clear` تمام آیتم‌ها را حذف می‌کند، ولی خود لیست (یک لیست خالی) باقی می‌ماند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
my_list.clear()
print(my_list)
```

<div dir="rtl">
<hr>

## حلقه زدن روی لیست

##### با استفاده از حلقه‌ی `for` می‌توانیم روی تک‌تک آیتم‌های یک لیست حرکت کنیم.

##### توضیحات کامل حلقه‌ها را در جلسه‌ی مربوط به شرط و حلقه‌ها می‌بینیم؛ اینجا فقط یک کاربرد ساده را می‌بینیم.

</div>


```python
my_list = ["apple", "banana", "cherry"]

for x in my_list:
    print(x)
```

<div dir="rtl">
<hr>

## مرتب‌سازی لیست — `()sort`

##### متد `()sort` لیست را به‌صورت **پیش‌فرض صعودی (حروف الفبا یا از کوچک به بزرگ)** مرتب می‌کند.

</div>


```python
my_list = ["orange", "mango", "kiwi", "pineapple", "banana"]
my_list.sort()
print(my_list)

numbers = [100, 50, 65, 82, 23]
numbers.sort()
print(numbers)
```

## 💻 تمرین کلاسی

##### کد زیر را کامل کنید:
#### ۱. عدد ۷ را با append به لیست اضافه کنید
#### ۲. لیست را با sort به‌صورت صعودی مرتب کنید
#### ۳. آخرین آیتم لیست را با pop حذف کنید
#### ۴. با یک حلقه for روی لیست حرکت کنید و هر عدد را چاپ کنید

</div>


```python
numbers = [5, 2, 9, 1]

numbers.___(7)
print(numbers)

numbers.___()
print(numbers)

numbers.___()
print(numbers)

___ n in numbers:
    print(n)
```

## جمع‌بندی

##### در این جلسه یاد گرفتیم:

- با ایندکس می‌توان مقدار یک آیتم را تغییر داد
- `append()` به انتها اضافه می‌کند، `insert()` در جایگاه دلخواه، `extend()` یک لیست دیگر را اضافه می‌کند
- `remove()` بر اساس مقدار حذف می‌کند، `pop()` بر اساس ایندکس، `del` هم آیتم و هم کل لیست را حذف می‌کند، `clear()` لیست را خالی می‌کند
- با `for` می‌توان روی آیتم‌های لیست حلقه زد
- `sort()` لیست را مرتب می‌کند
