## تاپل (Tuple) چیست؟

##### **تاپل** هم مثل لیست برای ذخیره‌ی چند مقدار در یک متغیر استفاده می‌شود، ولی با **پرانتز `()`** ساخته می‌شود.

##### تاپل‌ها هم **ترتیب‌دار** و **مجاز به تکرار** هستند، اما برخلاف لیست، **قابل تغییر نیستند (Unchangeable / Immutable)**.

</div>


```python
my_tuple = ("apple", "banana", "cherry")
print(my_tuple)
```

<div dir="rtl">

## ⚠️ نکته مهم — تاپل تک‌عضوی

##### برای ساخت یک تاپل با **فقط یک آیتم**، باید بعد از آن آیتم یک **کاما** بگذاریم؛ در غیر این صورت پایتون آن را تاپل نمی‌داند.

</div>


```python
my_tuple = ("apple",)
print(type(my_tuple))

# این یک تاپل نیست! این یک رشته است
test_tuple = ("apple")
print(type(test_tuple))
```

<div dir="rtl">

## دسترسی به آیتم‌های تاپل

##### دسترسی به آیتم‌های تاپل دقیقاً مثل لیست است: با **ایندکس** و **Slicing**.

</div>


```python
my_tuple = ("apple", "banana", "cherry", "orange", "kiwi")

print(my_tuple[1])     # ایندکس
print(my_tuple[-1])    # ایندکس منفی
print(my_tuple[1:3])   # Slicing
```

<div dir="rtl">
<hr>

## تغییر تاپل؟

##### همان‌طور که گفتیم، تاپل‌ها **Immutable** هستند و مستقیماً قابل تغییر نیستند.

##### اما یک راه‌حل داریم: تاپل را به **لیست** تبدیل می‌کنیم، تغییر می‌دهیم، و دوباره به تاپل برمی‌گردانیم.

</div>


```python
my_tuple = ("apple", "banana", "cherry")

y = list(my_tuple)     
y[1] = "kiwi" 
my_tuple = tuple(y)     

print(my_tuple)
```

<div dir="rtl">
<hr>

## Unpacking تاپل

##### می‌توانیم آیتم‌های یک تاپل را در **چند متغیر جداگانه** باز کنیم — به این کار **Unpacking** می‌گویند.

##### تعداد متغیرها باید با تعداد آیتم‌های تاپل برابر باشد.

</div>


```python
fruits = ("apple", "banana", "cherry")

(a, b, c) = fruits

print(a)
print(b)
print(c)
```

<div dir="rtl">
<hr>

## حلقه روی تاپل

##### دقیقاً مثل لیست، می‌توانیم با `for` روی آیتم‌های تاپل حلقه بزنیم.

</div>


```python
my_tuple = ("apple", "banana", "cherry")

for x in my_tuple:
    print(x)
```

## الحاق تاپل‌ها

##### با عملگر `+` می‌توانیم دو تاپل را به هم الحاق کنیم.

</div>


```python
tuple1 = ("a", "b", "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

<div dir="rtl">

## متدهای تاپل

##### تاپل چون قابل تغییر نیست، فقط **دو متد** دارد:

- `count()` ← تعداد تکرار یک مقدار را می‌شمارد
- `index()` ← جایگاه اولین وقوع یک مقدار را برمی‌گرداند

</div>


```python
my_tuple = (1, 3, 7, 8, 7, 3, 7)

print(my_tuple.count(7))
print(my_tuple.index(8))
```

## مجموعه (Set) چیست؟

##### **Set** برای ذخیره‌ی چند مقدار در یک متغیر استفاده می‌شود و با **آکولاد `{}`** ساخته می‌شود.

##### مجموعه‌ها سه ویژگی دارند که با لیست و تاپل فرق دارد:

- **بدون ترتیب (Unordered):** آیتم‌ها ترتیب مشخصی ندارند
- **بدون ایندکس (Unindexed):** نمی‌توان با ایندکس به آیتم‌ها دسترسی پیدا کرد
- **بدون تکرار (No Duplicates):** مقادیر تکراری به‌طور خودکار حذف می‌شوند

</div>


```python
my_set = {"apple", "banana", "cherry"}
print(my_set)
```

<div dir="rtl">

##### اگر مقدار تکراری در Set قرار دهیم، پایتون به‌طور خودکار یکی از آن‌ها را نگه می‌دارد.

</div>


```python
my_set = {"apple", "banana", "cherry", "apple"}
print(my_set)  
```

<div dir="rtl">

## بررسی وجود آیتم و حلقه روی Set

##### چون Set ایندکس ندارد، برای دسترسی به آیتم‌ها فقط می‌توان با `in` وجودشان را بررسی کرد یا با `for` روی آن‌ها حلقه زد.

</div>


```python
my_set = {"apple", "banana", "cherry"}

print("banana" in my_set)

#for x in my_set:
#    print(x)
```

<hr>

## افزودن آیتم به Set

##### برای افزودن **یک آیتم**، از `()add` و برای افزودن **چند آیتم** (یا محتوای یک مجموعه‌ی دیگر)، از `()update` استفاده می‌کنیم.

</div>


```python
my_set = {"apple", "banana", "cherry"}
my_set.add("orange")
print(my_set)

my_set.update(["mango", "grapes"])
print(my_set)
```

<div dir="rtl">
<hr>

## حذف آیتم از Set

##### با `()remove` یا `()discard` می‌توان یک آیتم را حذف کرد.

##### تفاوت مهم: اگر آیتم وجود نداشته باشد، `()remove` خطا می‌دهد، ولی `()discard` خطا نمی‌دهد.

</div>


```python
my_set = {"apple", "banana", "cherry"}

my_set.remove("banana")
print(my_set)

my_set.discard("melon")  # وجود ندارد، ولی خطا نمی‌دهد
print(my_set)
```
