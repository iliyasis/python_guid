## دیکشنری چیست؟

##### **دیکشنری (Dictionary)** برای ذخیره‌ی داده به‌صورت **جفت‌های کلید-مقدار (key:value)** استفاده می‌شود.

##### دیکشنری با **آکولاد `{}`** ساخته می‌شود.

##### ویژگی‌های دیکشنری:

- **ترتیب‌دار (Ordered):** از نسخه ۳.۷ به بعد، ترتیب درج آیتم‌ها حفظ می‌شود
- **قابل تغییر (Changeable):** می‌توان مقادیر را تغییر داد، اضافه یا حذف کرد
- **بدون کلید تکراری:** هر کلید فقط یک‌بار می‌تواند وجود داشته باشد

</div>


```python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)
```

<div dir="rtl">

##### طول دیکشنری (تعداد جفت‌های کلید-مقدار) هم با `len()` گرفته می‌شود.

</div>


```python
print(len(thisdict))
```

<div dir="rtl">
<hr>

## دسترسی به آیتم‌ها

##### برای دسترسی به یک مقدار، **نام کلید** را داخل کروشه می‌نویسیم، یا از متد `get()` استفاده می‌کنیم.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

x = thisdict["model"]
print(x)

y = thisdict.get("model")
print(y)
```

<div dir="rtl">

##### 🧠 نکته: تفاوت `[]` و `get()` این است که اگر کلید وجود نداشته باشد، `[]` **خطا** می‌دهد، ولی `get()` مقدار `None` (یا مقدار پیش‌فرضی که خودمان بدهیم) برمی‌گرداند.

</div>


```python
print(thisdict.get("color"))              # کلید وجود ندارد -> None
print(thisdict.get("color", "unknown"))   # مقدار پیش‌فرض دلخواه
```

<div dir="rtl">

## گرفتن کلیدها، مقادیر و هر دو

##### با `keys()` فقط کلیدها، با `values()` فقط مقادیر، و با `items()` هر دو (به‌صورت جفت) را می‌گیریم.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

print(thisdict.keys())
print(thisdict.values())
print(thisdict.items())
```

<div dir="rtl">

## بررسی وجود یک کلید — `in`

##### با `in` بررسی می‌کنیم آیا یک **کلید** در دیکشنری وجود دارد یا نه.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

if "model" in thisdict:
    print("بله، کلید 'model' وجود دارد")
```

<div dir="rtl">
<hr>

## تغییر مقادیر

##### برای تغییر مقدار یک کلید خاص، کافیست به همان کلید مقدار جدید بدهیم.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["year"] = 2018
print(thisdict)
```

<div dir="rtl">

## به‌روزرسانی با `update()`

##### متد `update()` مقدار یک یا چند کلید را همزمان به‌روزرسانی می‌کند؛ اگر کلید وجود نداشته باشد، آن را **اضافه** می‌کند.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

thisdict.update({"year": 2020})       # به‌روزرسانی
thisdict.update({"color": "red"})     # چون وجود نداشت، اضافه شد

print(thisdict)
```

<div dir="rtl">

## افزودن آیتم جدید

##### با نوشتن یک کلید جدید و مقدار آن، به‌سادگی یک آیتم جدید به دیکشنری اضافه می‌شود.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["color"] = "red"
print(thisdict)
```

<div dir="rtl">
<hr>

## حذف آیتم‌ها

##### چند روش برای حذف آیتم از دیکشنری داریم:

- `pop(key)` ← حذف با نام کلید
- `popitem()` ← حذف **آخرین** آیتم اضافه‌شده
- `del dict[key]` ← حذف با نام کلید (مثل pop)
- `del dict` ← حذف **کل دیکشنری**
- `clear()` ← خالی کردن دیکشنری

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

thisdict.pop("model")
print(thisdict)

thisdict.popitem()
print(thisdict)
```

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

del thisdict["model"]
print(thisdict)

thisdict.clear()
print(thisdict)  # دیکشنری خالی
```

<div dir="rtl">
<hr>

## حلقه روی دیکشنری

##### وقتی با `for` روی دیکشنری حلقه بزنیم، به‌صورت پیش‌فرض روی **کلیدها** حرکت می‌کند.

</div>


```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

# چاپ کلیدها
for x in thisdict:
    print(x)
```

<div dir="rtl">

##### برای گرفتن **مقادیر**، از `thisdict[x]` یا مستقیماً `thisdict.values()` استفاده می‌کنیم.

</div>


```python
for x in thisdict:
    print(thisdict[x])

for x in thisdict.values():
    print(x)
```

<div dir="rtl">

##### برای گرفتن **کلید و مقدار با هم**، از `items()` استفاده می‌کنیم.

</div>


```python
for key, value in thisdict.items():
    print(key, "-->", value)
```

## دیکشنری تودرتو (Nested Dictionary)

##### یک دیکشنری می‌تواند شامل **دیکشنری‌های دیگر** باشد؛ به این کار دیکشنری تودرتو می‌گویند.

##### این ساختار خیلی در دنیای واقعی (مثل داده‌های JSON) استفاده می‌شود.

</div>


```python
myfamily = {
    "child1": {
        "name": "Emil",
        "year": 2004
    },
    "child2": {
        "name": "Tobias",
        "year": 2007
    },
    "child3": {
        "name": "Linus",
        "year": 2011
    }
}

print(myfamily)

# دسترسی به یک مقدار در دیکشنری تودرتو
print(myfamily["child2"]["name"])
```

<div dir="rtl">

##### با یک حلقه‌ی `for` تودرتو می‌توانیم روی همه‌ی آیتم‌های یک دیکشنری تودرتو حرکت کنیم.

</div>


```python
for child, info in myfamily.items():
    print(child)
    for key in info:
        print("   ", key, "-->", info[key])
```

## ساخت دیکشنری با تابع سازنده — `dict()`

##### مثل بقیه‌ی انواع داده، دیکشنری هم یک تابع سازنده دارد. دقت کنید در این روش برای مقداردهی از `=` استفاده می‌شود، نه `:`.

</div>


```python
thisdict = dict(name="John", age=36, country="Norway")
print(thisdict)
```
