# تمرین — شرط‌ها و حلقه‌ها در پایتون

### سطح: مقدماتی

در این تمرین‌ها مباحث زیر را تمرین می‌کنیم:

- `if`, `elif`, `else`
- شرط‌های تودرتو
- حلقه `while`
- حلقه `for`
- تابع `range()`
- `break`
- `continue`
- ترکیب شرط‌ها و حلقه‌ها

**نکته:** سعی کن سؤال‌ها را بدون نگاه کردن به جواب حل کنی.

</div>

<div dir="rtl">

## 🟢 سطح ۱ — شرط‌ها

### تمرین ۱ — مثبت، منفی یا صفر

یک عدد از کاربر بگیر و مشخص کن:

- اگر مثبت بود: `Positive`
- اگر منفی بود: `Negative`
- اگر صفر بود: `Zero`

</div>

<div dir="rtl">

### تمرین ۲ — زوج یا فرد

یک عدد از کاربر بگیر و مشخص کن زوج است یا فرد.

مثال:

```text
Input: 17
Output: Odd
```

</div>

<div dir="rtl">

### تمرین ۳ — بزرگ‌ترین عدد

سه عدد از کاربر بگیر و **بدون استفاده از `max()`** بزرگ‌ترین آن‌ها را پیدا کن.

مثال:

```text
Input:
12
7
25

Output:
25
```

</div>

<div dir="rtl">

### تمرین ۴ — سیستم نمره‌دهی

نمره‌ی یک دانشجو را بگیر و طبق قوانین زیر رتبه را چاپ کن:

```text
18 تا 20   → A
15 تا 17   → B
12 تا 14   → C
10 تا 11   → D
کمتر از 10 → F
```

</div>

<div dir="rtl">

## 🟡 سطح ۲ — شرط + حلقه

### تمرین ۵ — شمارش اعداد زوج

با `for` و `range()` اعداد ۱ تا ۱۰۰ را بررسی کن:

1. اعداد زوج را چاپ کن.
2. در پایان تعداد اعداد زوج را چاپ کن.

**محدودیت:** از `len()` برای محاسبه تعداد استفاده نکن.

</div>

<div dir="rtl">

### تمرین ۶ — مجموع اعداد فرد

با استفاده از یک حلقه، مجموع اعداد فرد بین ۱ تا ۱۰۰ را حساب کن.

**محدودیت:** از `sum()` استفاده نکن.

</div>

<div dir="rtl">

### تمرین ۷ — شمارش مضرب‌ها

یک عدد از کاربر بگیر و تمام مضرب‌های آن عدد را بین ۱ تا ۱۰۰ چاپ کن.

مثلاً اگر ورودی `7` باشد:

```text
7
14
21
...
98
```

</div>

<div dir="rtl">

## 🟠 سطح ۳ — حلقه `while`

### تمرین ۸ — رمز عبور

رمز صحیح را در برنامه به صورت زیر قرار بده:

```python
password = 1234
```

از کاربر رمز بگیر.

تا زمانی که رمز اشتباه است، دوباره رمز را درخواست کن.

اگر درست بود:

```text
Access Granted
```

و اگر اشتباه بود:

```text
Wrong Password
```

**راهنمایی:** از `while` و `if` استفاده کن.

</div>

<div dir="rtl">

### تمرین ۹ — جمع کردن اعداد تا صفر

از کاربر عدد بگیر و اعداد واردشده را با هم جمع کن.

هر وقت کاربر `0` وارد کرد، دریافت اعداد متوقف شود و مجموع چاپ شود.

مثال:

```text
5
10
3
7
0

Sum: 25
```

</div>

<div dir="rtl">

### تمرین ۱۰ — بازی حدس عدد

عدد مخفی را در برنامه قرار بده:

```python
secret = 37
```

کاربر باید عدد را حدس بزند.

بعد از هر حدس:

- اگر حدس بزرگ‌تر بود → `Too High`
- اگر کوچک‌تر بود → `Too Low`
- اگر درست بود → `Correct!`

پس از حدس درست، برنامه تمام شود.

</div>

<div dir="rtl">

## 🔴 سطح ۴ — `break` و `continue`

### تمرین ۱۱ — اولین عدد بخش‌پذیر

اعداد ۱ تا ۱۰۰ را بررسی کن.

**اولین عددی که هم بر ۷ و هم بر ۵ بخش‌پذیر است** را پیدا کن و بعد با `break` حلقه را متوقف کن.

شرط توقف باید بر اساس پیدا شدن جواب باشد.

</div>

<div dir="rtl">

### تمرین ۱۲ — رد کردن اعداد

اعداد ۱ تا ۵۰ را چاپ کن، اما مضرب‌های ۳ چاپ نشوند.

**حتماً از `continue` استفاده کن.**

مثلاً:

```text
1
2
4
5
7
8
...
```

</div>

<div dir="rtl">

### تمرین ۱۳ — عدد اول

یک عدد از کاربر بگیر و مشخص کن **عدد اول هست یا نه**.

مثال:

```text
Input: 17
Output: Prime
```

و:

```text
Input: 18
Output: Not Prime
```

**راهنمایی:** با یک حلقه مقسوم‌علیه‌های عدد را بررسی کن.

</div>

<div dir="rtl">

## 🔥 سطح ۵ — تمرین‌های ترکیبی

### تمرین ۱۴ — ماشین حساب ساده

برنامه‌ای بنویس که دو عدد و یک عملگر از کاربر بگیرد.

عملگرهای مجاز:

```text
+
-
*
/
```

مثال:

```text
First number: 10
Operator: *
Second number: 5

Result: 50
```

اگر عملگر نامعتبر بود:

```text
Invalid Operator
```

</div>

<div dir="rtl">

### تمرین ۱۵ — منوی ساده

برنامه‌ای بنویس که دائماً این منو را نمایش دهد:

```text
1. Say Hello
2. Say Goodbye
3. Exit
```

اگر کاربر `1` زد:

```text
Hello!
```

اگر `2` زد:

```text
Goodbye!
```

اگر `3` زد، برنامه با `break` تمام شود.

در غیر این صورت:

```text
Invalid Choice
```

**راهنمایی:** از `while`، شرط‌ها و `break` استفاده کن.

</div>

<div dir="rtl">

### تمرین ۱۶ — گزارش اعداد

اعداد ۱ تا ۱۰۰ را بررسی کن و در پایان تعداد موارد زیر را چاپ کن:

```text
Even numbers: ...
Odd numbers: ...
Multiples of 5: ...
Multiples of 3 and 5: ...
```

**محدودیت:** از `sum()`، `count()` یا ابزارهای آماده برای شمارش استفاده نکن.

</div>

## 🏆 چالش نهایی

### تمرین ۱۷ — سیستم ورود ساده

یک سیستم Login بساز.

رمز صحیح:

```python
correct_password = 2468
```

کاربر حداکثر **۳ بار** فرصت دارد رمز را وارد کند.

اگر درست وارد کرد:

```text
Login Successful
```

و برنامه تمام شود.

اگر سه بار اشتباه کرد:

```text
Account Locked
```

### شرایط

- از `while` استفاده کن.
- تعداد تلاش‌ها را بشمار.
- از `if / else` استفاده کن.
- در صورت ورود موفق، با `break` از حلقه خارج شو.

</div>

بعد از هر حدس:

# 🎯 چک‌لیست پایان تمرین

بعد از حل تمرین‌ها باید بتوانی:

- [ ] با `if / elif / else` تصمیم‌گیری کنی.
- [ ] شرط‌های چندگانه بنویسی.
- [ ] با `for` روی `range()` حرکت کنی.
- [ ] با `while` یک فرایند را تا برقرار بودن یک شرط تکرار کنی.
- [ ] از `break` برای خروج از حلقه استفاده کنی.
- [ ] از `continue` برای رد کردن یک تکرار استفاده کنی.
- [ ] شرط و حلقه را با هم ترکیب کنی.
- [ ] یک مسئله‌ی ساده را بدون دیدن کد نمونه به برنامه تبدیل کنی.

**پیشنهاد:** اگر در یک تمرین گیر کردی، اول مسئله را روی کاغذ به چند مرحله‌ی کوچک تقسیم کن و بعد هر مرحله را به کد تبدیل کن.
