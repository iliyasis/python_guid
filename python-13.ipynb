## دستور شرطی `if`

##### پایتون از کلمات کلیدی `if`, `elif`, `else` برای شرط‌گذاری استفاده می‌کند.

##### به یاد بیاورید: پایتون برای مشخص کردن بلوک کد از **تورفتگی (Indentation)** استفاده می‌کند، نه آکولاد.
####  قبل از هر چیز بیاید با ورودی گرفتن در پایتون آشنا بشیم
</div>


```python
#input!!!!!
```

```python
a = 33
b = 200

if b > a:
    print("b is greater than a")
```

<div dir="rtl">

## `elif`

##### `elif` یعنی "اگر شرط‌های قبلی درست نبودند، این شرط را امتحان کن".

</div>


```python
a = 33
b = 33

if b > a:
    print("b is greater than a")
elif a == b:
    print("a is equal to b")
```

<div dir="rtl">

## `else`

##### `else` باقی‌مانده‌ی شرط را می‌گیرد؛ یعنی اگر هیچ‌کدام از شرط‌های بالا درست نبودند.

</div>


```python
a = 200
b = 33

if b > a:
    print("b is greater than a")
elif a == b:
    print("a is equal to b")
else:
    print("a is greater than b")
```

## شرط تودرتو (Nested If)

##### می‌توانیم یک `if` را داخل `if` دیگری بنویسیم.

</div>


```python
x = 41

if x > 10:
    print("bishtar az 10")
    if x > 20:
        print("bishtar az 20")
    else:
        print("bishtar az 10 ammmma! kamtar az 20")
```

## شرط تک‌خطی (Short Hand)

##### اگر بدنه‌ی شرط فقط **یک دستور** باشد، می‌توان آن را در یک خط نوشت.

</div>


```python
a = 200
b = 33

if a > b: print("a is greater")
```

<div dir="rtl">

##### همچنین برای انتخاب بین دو مقدار در **یک خط**، می‌توان از یک ساختار شرطی فشرده استفاده کرد که به آن **Ternary Operator** هم می‌گویند.

</div>


```python
a = 2
b = 330

print("A") if a > b else print("B")
```

## حلقه `while`

##### حلقه‌ی `while` تا زمانی که شرطش **درست (True)** باشد، بدنه‌اش را تکرار می‌کند.

</div>


```python
i = 1

while i < 6:
    print(i)
    i = i + 1
```

<div dir="rtl">

##### ⚠️ نکته بسیار مهم: اگر فراموش کنیم مقدار متغیر شرط را در بدنه‌ی حلقه تغییر دهیم، حلقه **هیچ‌وقت تمام نمی‌شود** (حلقه‌ی بی‌نهایت / Infinite Loop).

</div>

## حلقه `for`

##### حلقه‌ی `for` برای **حرکت روی یک دنباله** (لیست، تاپل، دیکشنری، Set یا رشته) استفاده می‌شود.

##### همان‌طور که در جلسات قبل دیدیم، با `for` می‌توانیم روی رشته، لیست، تاپل، Set و دیکشنری حلقه بزنیم.

</div>


```python
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    print(x)
```

<div dir="rtl">

## تابع `()range`

##### برای حلقه زدن روی مجموعه‌ای از **اعداد**، از تابع `()range` استفاده می‌کنیم.

##### `range(6)` یعنی اعداد ۰ تا ۵ (شامل ۰، ولی نه شامل ۶).

</div>


```python
for x in range(6):
    print(x)
```

<div dir="rtl">

##### `()range` می‌تواند نقطه‌ی شروع، پایان و **گام (Step)** هم بگیرد:

</div>


```python
for x in range(2, 30, 3):
    print(x)
```

## حلقه‌ی تودرتو (Nested Loop)

##### یک حلقه داخل حلقه‌ی دیگر؛ به‌ازای هر بار اجرای حلقه‌ی بیرونی، حلقه‌ی داخلی **کامل** اجرا می‌شود.

</div>


```python
halat = ["fast", "big", "small"]
fruits = ["apple", "banana", "cherry"]

for x in halat:
    for y in fruits:
        print(x, y)
```

## `break` — خروج از حلقه

##### دستور `break` باعث می‌شود حلقه **قبل از پایان طبیعی‌اش متوقف** شود.

</div>


```python
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    print(x)
    if x == "banana":
        break
```

<div dir="rtl">

## `continue` — رد کردن یک تکرار

##### دستور `continue` باقی‌مانده‌ی این دور از حلقه را رد می‌کند و به **تکرار بعدی** می‌رود.

</div>


```python
#mesal adadi!!!!!!!
fruits = ["apple", "banana", "cherry"]

for x in fruits:
    if x == "banana":
        continue
    print(x)
```

<div dir="rtl">

##### `break`، `continue`  دقیقاً به همین شکل در حلقه‌ی `while` هم قابل استفاده‌اند.

</div>


```python
i = 0

while i < 6:
    i += 1
    if i == 3:
        continue
    if i == 5:
        break
    print(i)
```
