## تعریف تابع

##### **تابع (Function)** یک بلوک کد است که فقط زمانی اجرا می‌شود که آن را **فراخوانی** کنیم.

##### در پایتون تابع با کلمه‌کلیدی `def` تعریف می‌شود.

</div>


```python
def my_function():
    print("hello")
```

<div dir="rtl">

## فراخوانی تابع

##### برای اجرای یک تابع، نام آن را همراه با پرانتز `()` می‌نویسیم.

</div>


```python
def my_function():
    print("hello")

my_function()
```

<div dir="rtl">
<hr>

## پارامتر و آرگومان

##### اطلاعات را می‌توان به‌صورت **پارامتر (Parameter)** به تابع پاس داد.

##### پارامتر داخل پرانتز، هنگام **تعریف** تابع نوشته می‌شود؛ آرگومان مقداری است که هنگام **فراخوانی** به آن پارامتر می‌دهیم.

</div>


```python
def my_function(f_name):
    print(f_name + " sisakhti")

my_function("Ali")
my_function("Sara")
my_function("shirin")
```

<div dir="rtl">

## تعداد آرگومان‌ها باید درست باشد

##### به‌طور پیش‌فرض، یک تابع باید با **همان تعداد آرگومانی** که در تعریف مشخص شده، فراخوانی شود.

</div>


```python
def my_function(f_name, l_name):
    print(f_name + " " + l_name)

my_function("Ali", "sisakhti")

# my_function("Ali")
```

<div dir="rtl">

## آرگومان‌های کلیدی (Keyword Arguments)

##### می‌توانیم آرگومان‌ها را به‌صورت `کلید=مقدار` بفرستیم. در این حالت **ترتیب اهمیتی ندارد**.

</div>


```python
def my_function(f_name, l_name):
    print(f_name + " " + l_name)

my_function(l_name="sisakhti", f_name="Ali")
```

<div dir="rtl">
<hr>

## آرگومان‌های نامحدود — `args*`

##### اگر تعداد دقیق آرگومان‌ها را از قبل ندانیم، می‌توانیم قبل از نام پارامتر یک `*` بگذاریم.

##### در این حالت تابع همه‌ی آرگومان‌ها را در یک **تاپل** دریافت می‌کند.

</div>


```python
def my_function(*kids):
    print(kids)

my_function("Emil", "Tobias", "Linus")
```

<div dir="rtl">

## آرگومان‌های کلیدی نامحدود — `kwargs**`

##### اگر تعداد آرگومان‌های **کلیدی** را ندانیم، از `**` قبل از نام پارامتر استفاده می‌کنیم.

##### در این حالت تابع همه‌ی آرگومان‌ها را در یک **دیکشنری** دریافت می‌کند.

</div>


```python
def my_function(**kid):
    print(kid)

my_function(fname="Tobias", lname="Refsnes")
```

<div dir="rtl">
<hr>

## مقدار پیش‌فرض پارامتر

##### می‌توانیم برای یک پارامتر **مقدار پیش‌فرض** تعیین کنیم؛ اگر هنگام فراخوانی مقداری برای آن ندهیم، همان مقدار پیش‌فرض استفاده می‌شود.

</div>


```python
def my_function(country="Norway"):
    print("i am from" , country)

my_function("Sweden")
#my_function("India")
#my_function()
#my_function("Brazil")
```

<div dir="rtl">

## ارسال لیست به‌عنوان آرگومان

##### می‌توانیم هر نوع داده‌ای (رشته، عدد، لیست، دیکشنری و...) را به‌عنوان آرگومان بفرستیم؛ نوع آن داخل تابع هم حفظ می‌شود.

</div>


```python
def my_function(food):
    for x in food:
        print(x)

fruits = ["apple", "banana", "cherry"]
my_function(fruits)
```

<div dir="rtl">
<hr>

## دستور `return`

##### با `return` می‌توانیم از داخل تابع یک **مقدار** را به بیرون برگردانیم.

##### به محض رسیدن به `return`، تابع بلافاصله **متوقف** می‌شود.

</div>


```python
def my_function(x):
    return 5 * x

print(my_function(3))
print(my_function(5))
print(my_function(9))
```

<div dir="rtl">
<hr>

## محدوده‌ی متغیرها (Scope)

##### متغیری که **داخل یک تابع** تعریف می‌شود، یک متغیر **محلی (Local)** است و فقط داخل همان تابع در دسترس است.

</div>


```python
def myfunc():
    x = 300   # local variable
    print(x)

myfunc()

# print(x)
```

<div dir="rtl">

## متغیر سراسری (Global)

##### متغیری که **بیرون از هر تابعی** تعریف شود، یک متغیر **سراسری (Global)** است و در همه‌جا از جمله داخل توابع قابل خواندن است.

</div>


```python
x = 300   # global variable

def myfunc():
    print(x)

myfunc()
print(x)
```

<div dir="rtl">

##### ⚠️ اگر داخل تابع متغیری با **همان نام** بسازیم، پایتون آن را یک متغیر **محلی جدید** می‌داند، نه تغییر متغیر سراسری.

##### برای تغییر واقعی متغیر سراسری از داخل تابع، باید از کلمه‌کلیدی `global` استفاده کنیم.

</div>


```python
x = 300

def myfunc():
    global x 
    x = 200  

myfunc()
print(x)
```

## Lambda — توابع بی‌نام

##### **Lambda** یک تابع کوچک و بی‌نام است که می‌تواند **هر تعداد آرگومان** ولی فقط **یک عبارت (Expression)** داشته باشد.

</div>


```python
x = lambda a: a + 10
print(x(5))
```

<div dir="rtl">

##### Lambda می‌تواند چند آرگومان هم داشته باشد.

</div>


```python
x = lambda a, b: a * b
print(x(5, 6))

x = lambda a, b, c: a + b + c
print(x(5, 6, 2))
```
