## `try` و `except`

##### با `try` و `except` می‌توانیم این خطاها را **مدیریت** کنیم تا برنامه متوقف نشود.

##### بلوک `try` کدی را که ممکن است خطا بدهد اجرا می‌کند؛ بلوک `except` وقتی خطا رخ دهد اجرا می‌شود.

</div>


```python
try:
    print(x)
except:
    print("error")
```

<div dir="rtl">

## مدیریت انواع مختلف خطا

##### می‌توانیم برای **انواع خاصی از خطا** بلوک `except` جداگانه بنویسیم.

</div>


```python
try:
    print(x)
except NameError:
    print("x tarif nashode")
except:
    print("error")
```

<div dir="rtl">

##### چند نوع خطای پرکاربرد دیگر: `ZeroDivisionError` (تقسیم بر صفر)، `ValueError` (مقدار نامعتبر برای تبدیل)، `TypeError` (نوع داده‌ی نامناسب).

</div>


```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Division by zero is impossible.")

try:
    number = int("abc")
except ValueError:
    print("این رشته قابل تبدیل به عدد نیست")
```

<div dir="rtl">
<hr>

## بند `else`

##### بلوک `else` وقتی اجرا می‌شود که در `try` هیچ خطایی رخ **نداده** باشد.

</div>


```python
try:
    print("hello")
except:
    print("error")
else:
    print("good")
```

<div dir="rtl">

## بند `finally`

##### بلوک `finally`، **در هر صورت** (چه خطا رخ دهد چه ندهد) اجرا می‌شود.

##### این بلوک معمولاً برای **پاک‌سازی منابع** استفاده می‌شود؛ مثلاً بستن یک فایل که باز کرده‌ایم.

</div>


```python
try:
    print(x)
except:
    print("A problem")
finally:
    print("tamam shod")
```

## دستور `raise`

##### اگر خودمان بخواهیم عمداً یک خطا **صادر (Raise)** کنیم — مثلاً برای بررسی یک شرط مهم — از `raise` استفاده می‌کنیم.

</div>


```python
x = -1

if x < 0:
    raise Exception("عدد نمی‌تواند منفی باشد")
```

## گرفتن ورودی از کاربر — `input()`

##### پایتون با تابع `input()` اجازه می‌دهد از کاربر **ورودی** بگیریم.

##### پایتون در اجرای برنامه متوقف می‌شود و منتظر می‌ماند تا کاربر چیزی وارد کند.

</div>


```python
username = input("enter your user name: ")
print("your username: " + username)
```

<div dir="rtl">

## ⚠️ نکته مهم — ورودی همیشه رشته است

##### مقدار برگشتی `input()` همیشه از نوع **رشته (str)** است، حتی اگر کاربر یک عدد وارد کند.

##### برای انجام محاسبات، باید ورودی را با `int()` یا `float()` به عدد تبدیل کنیم.

</div>


```python
age_text = input("enter your age: ")
print(type(age_text))   

age = int(age_text)     
print(type(age))       
print("دو سال دیگر شما " + str(age + 2) + " سال دارید")
```

<div dir="rtl">

##### ترکیب `input()` با `try/except` برای گرفتن **ورودی امن و معتبر** بسیار پرکاربرد است.

</div>


```python
try:
    age = int(input("enter your age: "))
    print("age: ", age)
except ValueError:
    print("yek adad vared kon: ")
```

## مرور و تکمیل قالب‌بندی رشته (F-String)

##### در جلسه‌ی رشته‌ها با F-String آشنا شدیم. یادآوری: با گذاشتن `f` قبل از رشته و مقدار داخل `{}` می‌توانیم متغیر و عدد را ترکیب کنیم.

</div>


```python
name = "Ali"
age = 25
print(f"name: {name}, age: {age}")
```

<div dir="rtl">

## Modifier در F-String

##### با Modifier می‌توانیم قالب عدد را دقیق‌تر مشخص کنیم: تعداد رقم اعشار، جداکننده‌ی هزارگان، درصد و...

</div>


```python
price = 59.9999
print(f"price: {price:.2f}")     # ۲ رقم اعشار

big_number = 1234567
print(f"number: {big_number:,}")   # جداکننده‌ی هزارگان

rate = 0.256
print(f"darsad: {rate:.1%}")      # نمایش به‌صورت درصد
```

## مقدمه‌ای بر ماژول‌ها (Modules)

##### **ماژول** فایلی حاوی کدهای پایتون (متغیر، تابع، کلاس) است که می‌توانیم آن را در برنامه‌ی خود **import** کنیم.

##### پایتون تعداد زیادی ماژول **آماده و داخلی (Built-in)** دارد؛ مثل `random` که در جلسه‌ی اعداد دیدیم.

</div>


```python
import math

print(math.sqrt(16))     # جذر
print(math.pi)           # عدد پی
print(math.floor(4.7))   # گرد کردن به پایین
```

<div dir="rtl">

##### با `from ماژول import نام` می‌توانیم فقط یک بخش خاص از ماژول را وارد کنیم و بدون نوشتن نام ماژول از آن استفاده کنیم.

</div>


```python
from math import sqrt

print(sqrt(25))
```
