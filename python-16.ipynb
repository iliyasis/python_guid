## تابع `()open`

##### تابع اصلی برای کار با فایل، `()open` است که دو پارامتر می‌گیرد: **نام فایل** و **حالت (mode)**.

##### چهار حالت اصلی:

| حالت | توضیح |
|---|---|
| `"r"` | خواندن (پیش‌فرض) — اگر فایل نباشد خطا می‌دهد |
| `"w"` | نوشتن — اگر فایل نباشد می‌سازد، اگر باشد **محتوایش را پاک می‌کند** |
| `"a"` | افزودن — اگر فایل نباشد می‌سازد، در غیر این صورت به انتها اضافه می‌کند |
| `"x"` | ایجاد — اگر فایل از قبل وجود داشته باشد خطا می‌دهد |

</div>

## ساخت و نوشتن یک فایل

##### ابتدا با حالت `"w"` یک فایل نمونه می‌سازیم تا در ادامه با آن کار کنیم.

</div>


```python
f = open("demofile.txt", "w")
f.write("satr aval\n")
f.write("satre dovom\n")
f.close() 

print("file sakhte shod")
```

<div dir="rtl">

##### ⚠️ نکته مهم: حالت `"w"` اگر فایل از قبل وجود داشته باشد، **کل محتوای آن را پاک می‌کند**! برای اضافه کردن بدون پاک کردن، از حالت `"a"` استفاده می‌کنیم.

</div>


```python
f = open("demofile.txt", "a")
f.write("khat jadid ezafe shod\n")
f.close()
```

## خواندن فایل

##### با متد `read()` می‌توانیم **کل محتوای** فایل را به‌صورت یک رشته بخوانیم.

</div>


```python
f = open("demofile.txt", "r")
print(f.read())
f.close()
```

<div dir="rtl">

##### با `read(n)` می‌توانیم فقط **n کاراکتر اول** را بخوانیم.

</div>


```python
f = open("demofile.txt", "r")
print(f.read(10))   # فقط ۱۰ کاراکتر اول
f.close()
```

<div dir="rtl">

## خواندن خط به خط — `readline()`

##### با `readline()` می‌توانیم فقط **یک خط** از فایل را بخوانیم؛ هر بار که آن را صدا بزنیم، خط بعدی را می‌خواند.

</div>


```python
f = open("demofile.txt", "r")
print(f.readline())   # خط اول
print(f.readline())   # خط دوم
f.close()
```

<div dir="rtl">

## حلقه روی خطوط فایل

##### با یک حلقه‌ی `for` می‌توانیم روی **تمام خطوط** فایل حرکت کنیم — رایج‌ترین روش خواندن فایل.

</div>


```python
f = open("demofile.txt", "r")
for line in f:
    print(line, end="")
f.close()
```

<div dir="rtl">
<hr>

## دستور `with` — روش امن‌تر

##### روش پیشنهادی و مدرن برای کار با فایل، استفاده از `with` است.

##### مزیت اصلی: فایل به‌طور **خودکار بسته می‌شود**، حتی اگر داخل بلوک خطایی رخ دهد — دیگر نیازی به `close()` نیست.

</div>


```python
with open("demofile.txt", "r") as f:
    content = f.read()
    print(content)

# اینجا f از قبل بسته شده است، حتی بدون نوشتن f.close()
```

```python
with open("demofile.txt", "a") as f:
    f.write("ba ravesh with")
```

## حذف فایل — ماژول `os`

##### پایتون خودش تابعی برای حذف فایل ندارد؛ باید ماژول `os` را وارد کنیم.

##### با `os.path.exists()` هم می‌توانیم قبل از حذف، وجود فایل را بررسی کنیم تا خطا نگیریم.

</div>


```python
import os

if os.path.exists("demofile.txt"):
    os.remove("demofile.txt")
    print("deleted")
else:
    print("not found")
```

## کار با JSON

##### **JSON** یک فرمت سبک و خوانا برای **ذخیره و تبادل داده** است که خیلی شبیه دیکشنری پایتون است.

##### برای کار با آن، ماژول داخلی `json` را وارد می‌کنیم.

</div>


<div dir="rtl">

## تبدیل پایتون به JSON — `json.dumps()`

##### با `json.dumps()` یک شیء پایتون (مثل دیکشنری) را به یک **رشته‌ی JSON** تبدیل می‌کنیم.

</div>


```python
import json

person = {
    "name": "John",
    "age": 30,
    "city": "New York",
    "married": True,
    "children": ["Ann", "Billy"]
}

json_string = json.dumps(person)
print(json_string)
print(type(json_string))
```

<div dir="rtl">

##### با پارامتر `indent` می‌توان خروجی JSON را **خواناتر** و با تورفتگی چاپ کرد.

</div>


```python
print(json.dumps(person, indent=4))
```

<div dir="rtl">

## تبدیل JSON به پایتون — `json.loads()`

##### برعکس، با `json.loads()` یک رشته‌ی JSON را به یک **دیکشنری پایتون** تبدیل می‌کنیم.

</div>


```python
x = '{ "name":"John", "age":30, "city":"New York"}'

y = json.loads(x)
print(y)
print(y["age"])
```

<div dir="rtl">

## ذخیره و خواندن JSON در فایل — `json.dump()` و `json.load()`

##### برای نوشتن مستقیم داده در یک **فایل JSON** از `json.dump()` و برای خواندن از `json.load()` استفاده می‌کنیم.

##### دقت کنید: `dumps`/`loads` با **رشته** کار می‌کنند، `dump`/`load` (بدون s) با **فایل** کار می‌کنند.

</div>


```python
data = {"name": "Alice", "age": 28, "city": "Tehran"}

# نوشتن در فایل JSON
with open("data.json", "w") as f:
    json.dump(data, f, indent=4)

# خواندن از فایل JSON
with open("data.json", "r") as f:
    loaded = json.load(f)

print(loaded)
print(loaded["name"])
```

## کار با CSV

##### **CSV (Comma-Separated Values)** فرمتی بسیار رایج برای ذخیره‌ی داده‌های **جدولی** است (مثل خروجی اکسل).

##### هر خط یک ردیف، و مقادیر با **کاما** از هم جدا می‌شوند. برای کار با آن، ماژول `csv` را وارد می‌کنیم.

</div>


<div dir="rtl">

## نوشتن فایل CSV — `csv.writer`

</div>


```python
import csv

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age", "Major"])       # سرستون‌ها
    writer.writerow(["Ali", 20, "Computer Science"])
    writer.writerow(["Sara", 22, "Mathematics"])

print("فایل CSV ساخته شد")
```

<div dir="rtl">

## خواندن فایل CSV — `csv.reader`

##### هر ردیف با `csv.reader` به‌صورت یک **لیست** برگردانده می‌شود.

</div>


```python
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

<div dir="rtl">

## کار با CSV به‌صورت دیکشنری — `DictReader` / `DictWriter`

##### با `DictReader` هر ردیف به‌صورت یک **دیکشنری** (بر اساس نام ستون‌ها) خوانده می‌شود که کار با آن راحت‌تر است.

</div>


```python
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Name"], "-->", row["Major"])
```
