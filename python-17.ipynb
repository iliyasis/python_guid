## کلاس (Class) چیست؟

##### تقریباً همه‌چیز در پایتون یک شیء (Object) است، با ویژگی‌ها (Properties) و متدها (Methods) خودش.

##### **کلاس** یک **قالب (Template)** برای ساختن شیء است.

##### کلاس با کلمه‌کلیدی `class` تعریف می‌شود.

</div>


```python
class Person:
    name = "John"
    age = 36
```

<div dir="rtl">
<hr>

## شیء (Object) چیست؟

##### با استفاده از کلاس می‌توانیم **شیء (Object)** بسازیم — به این کار **نمونه‌سازی (Instantiation)** می‌گویند.

</div>


```python
class Person:
    name = "John"
    age = 36

p1 = Person()  
print(p1.name)
print(p1.age)
```

<div dir="rtl">
<hr>

## متد `()__init__`

##### همه‌ی کلاس‌ها یک متد ویژه به نام `()__init__` دارند که همیشه هنگام **ساخته شدن شیء جدید** به‌طور خودکار اجرا می‌شود.

##### از `()__init__` برای مقداردهی اولیه به **ویژگی‌های شیء** استفاده می‌کنیم.

</div>


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)

print(p1.name)
print(p1.age)
```

<div dir="rtl">

##### 🧠 نکته: متد `()__init__` برای هر شیء **جدید** به‌طور خودکار فراخوانی می‌شود؛ خودمان مستقیماً آن را صدا نمی‌زنیم.

</div>


<div dir="rtl">
<hr>

## پارامتر `self`

##### پارامتر `self` به **همان شیء فعلی** اشاره می‌کند و برای دسترسی به ویژگی‌هایی که به کلاس تعلق دارند استفاده می‌شود.

##### `self` باید **همیشه اولین پارامتر** هر متد در کلاس باشد؛ بدون آن، پایتون نمی‌داند کدام شیء مورد نظر است.

</div>


```python
class Person:
    def __init__(self, name):
        self.name = name

    def printname(self):
        print(self.name)

p1 = Person("Tobias")
p2 = Person("Linus")

p1.printname()
p2.printname()
```

<div dir="rtl">

##### 📌 نکته جالب: نام `self` اجباری نیست (می‌توانستیم هر اسم دیگری بگذاریم)، ولی طبق **قرارداد رایج** همیشه `self` نامیده می‌شود و بهتر است همین قرارداد رعایت شود.

</div>


<div dir="rtl">
<hr>

## متدهای شیء (Object Methods)

##### شیءها می‌توانند **متد** هم داشته باشند — یعنی توابعی که به شیء تعلق دارند.

</div>


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def myfunc(self):
        print("hello my name is " + self.name)

p1 = Person("John", 36)
p1.myfunc()
```

<div dir="rtl">

##### متدها می‌توانند مثل توابع معمولی مقداری هم **برگردانند (return)**.

</div>


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_info(self):
        return f"{self.name} is {self.age} years old"

p1 = Person("Tobias", 28)
print(p1.get_info())
```

<hr>

## تغییر ویژگی‌های شیء

##### می‌توانیم مقدار یک ویژگی را روی یک شیء خاص **تغییر** دهیم.

</div>


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)

p1.age = 40
print(p1.age)
```

<div dir="rtl">

## حذف ویژگی و حذف شیء

##### با `del` می‌توانیم یک **ویژگی خاص** یا حتی **کل شیء** را حذف کنیم.

</div>


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)

del p1.age    # حذف فقط ویژگی age
print(p1.name)

# این خطا می‌دهد چون age حذف شده:
# print(p1.age)
```

```python
p1 = Person("John", 36)
del p1   # حذف کل شیء

# این خطا می‌دهد چون p1 دیگر وجود ندارد:
# print(p1.name)
```

## مثال کامل — کلاس با چند ویژگی و متد

##### یک مثال جامع‌تر که چند ویژگی و چند متد را با هم ترکیب می‌کند.

</div>


```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        self.masafat = 0   # مقدار پیش‌فرض، وابسته به ورودی نیست

    def drive(self, km):
        self.masafat += km
        print(f"{self.brand}  + {km} kilometer")

    def show_info(self):
        print(f"{self.brand} {self.model} ({self.year}) and {self.masafat} kilometer")


car1 = Car("Toyota", "Corolla", 2022)
car1.drive(150)
car1.drive(80)
car1.show_info()
```

</div>


```python
class Book:
    def ___(self, title, author, price):
        self.title = title
        self.author = ___
        self.price = price

    def show(___):
        print(f"{self.title} by {self.author} - ${self.price}")

b1 = Book("Python 101", "John Smith", 25)
b1.___()

b1.___ = 20
b1.show()
```
