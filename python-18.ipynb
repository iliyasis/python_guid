## وراثت (Inheritance) چیست؟

##### **وراثت** به ما اجازه می‌دهد یک کلاس جدید (**کلاس فرزند / Child**) بسازیم که تمام ویژگی‌ها و متدهای یک کلاس دیگر (**کلاس والد / Parent**) را به ارث می‌برد.

##### این کار باعث می‌شود کد **کمتر تکراری** و **قابل استفاده مجدد** باشد.

</div>


<div dir="rtl">

## ساخت کلاس والد (Parent Class)

##### هر کلاس معمولی می‌تواند به‌عنوان کلاس والد استفاده شود.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

x = Person("John", "Doe")
x.printname()
```

## ساخت کلاس فرزند (Child Class)

##### برای ساخت کلاس فرزند، نام کلاس والد را داخل پرانتز جلوی نام کلاس فرزند می‌نویسیم.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)


class Student(Person):
    pass

x = Student("Mike", "Olsen")
x.printname()  
```

##### همان‌طور که دیدید، کلاس `Student` بدون اینکه چیزی بنویسیم، **همه‌ی ویژگی‌ها و متدهای** کلاس `Person` را دارد.

</div>

## اضافه کردن `()__init__` به کلاس فرزند

##### وقتی کلاس فرزند متد `()__init__` خودش را اضافه کند، دیگر به‌طور خودکار `()__init__` والد را اجرا **نمی‌کند** — باید خودمان آن را صدا بزنیم.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)



class Student(Person):
    def __init__(self, fname, lname):
        Person.__init__(self, fname, lname)

x = Student("Mike", "Olsen")
x.printname()
```

## تابع `()super`

##### پایتون تابعی به نام `()super` دارد که کار بالا را **ساده‌تر** می‌کند؛ دیگر نیازی نیست نام کلاس والد را بنویسیم.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

        


class Student(Person):
    def __init__(self, fname, lname):
        super().__init__(fname, lname) 

x = Student("Mike", "Olsen")
x.printname()
```

## اضافه کردن ویژگی و متد جدید

##### در کلاس فرزند می‌توانیم علاوه‌بر آنچه از والد به ارث برده‌ایم، **ویژگی و متد جدید** هم اضافه کنیم.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

        

class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)
        self.faregh = year

    def welcome(self):
        print("welcom to university", self.firstname, self.lastname, "sal fargh :", self.faregh)


x = Student("Mike", "Olsen", 2025)
x.printname()
x.welcome()
```

## Override کردن متد والد

##### اگر در کلاس فرزند متدی با **همان نام** متد والد بنویسیم، نسخه‌ی فرزند جایگزین (Override) می‌شود.

</div>


```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)




class Student(Person):
    def __init__(self, fname, lname):
        super().__init__(fname, lname)

    def printname(self):
        print("student:", self.firstname, self.lastname)


x = Student("Mike", "Olsen")
x.printname()   
```

## پلی‌مورفیسم (Polymorphism)

##### **پلی‌مورفیسم** یعنی «چند شکل بودن» — یعنی کلاس‌های مختلف می‌توانند متدی با **نام یکسان** داشته باشند که هرکدام رفتار متفاوتی دارد.

##### به‌همین دلیل می‌توانیم همان متد را روی شیءهایی از کلاس‌های مختلف، بدون دانستن نوع دقیق شیء، صدا بزنیم.

</div>


```python
class Cat:
    def make_sound(self):
        print("mio mio")

class Dog:
    def make_sound(self):
        print("hap hap")

class Cow:
    def make_sound(self):
        print("maaaa maaaa")


animals = [Cat(), Dog(), Cow()]

for animal in animals:
    animal.make_sound()
```

## کپسوله‌سازی (Encapsulation)

##### **کپسوله‌سازی** یعنی محدود کردن دسترسی مستقیم به بعضی از ویژگی‌های یک کلاس، برای **محافظت از داده**.

##### در پایتون سه سطح دسترسی داریم:

| نوع | نوشتار | دسترسی |
|---|---|---|
| عمومی (Public) | `name` | از همه‌جا قابل دسترسی |
| محافظت‌شده (Protected) | `name_` | فقط یک قرارداد؛ یعنی «فقط داخل کلاس/فرزندان استفاده شود» |
| خصوصی (Private) | `name__` | از بیرون کلاس مستقیماً قابل دسترسی نیست |

</div>


```python
class BankAccount:
    def __init__(self, owner, mojodi):
        self.owner = owner
        self.__mojodi = mojodi   # ویژگی خصوصی (private)

account = BankAccount("Ali", 1000)
print(account.owner)

# این خطا می‌دهد چون __mojodi خصوصی است:
# print(account.__mojodi)
```

## Getter و Setter

##### چون به ویژگی‌های خصوصی مستقیماً دسترسی نداریم، از **متدهای Getter و Setter** برای خواندن و تغییر امن آن‌ها استفاده می‌کنیم.

</div>


```python
class BankAccount:
    def __init__(self, owner, mojodi):
        self.owner = owner
        self.__mojodi = mojodi

    def get_balance(self):
        return self.__mojodi

    def deposit(self, amount):
        self.__mojodi += amount


account = BankAccount("Ali", 1000)
print(account.get_balance())

account.deposit(500)
print(account.get_balance())
```
