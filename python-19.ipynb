## ۱. پایگاه داده چیست؟

پایگاه داده (**Database**) محلی برای ذخیره و مدیریت اطلاعات است.

مثلاً برای اطلاعات دانش‌آموزان می‌توانیم جدولی مثل این داشته باشیم:

| id | name | age | score |
|---:|---|---:|---:|
| 1 | علی | 16 | 18.5 |
| 2 | سارا | 17 | 19 |
| 3 | رضا | 16 | 15.5 |

در جدول:

- هر **ستون (Column)** یک نوع اطلاعات را مشخص می‌کند.
- هر **ردیف (Row)** اطلاعات یک رکورد است.
- `id` می‌تواند شناسه‌ی یکتای هر رکورد باشد.

</div>

## ۲. SQLite چیست؟

**SQLite** یک سیستم مدیریت پایگاه داده‌ی سبک و فایل‌محور است.

## ۷. مرحله سوم: ساخت Cursor

برای فرستادن دستورات SQL به دیتابیس، یک Cursor می‌سازیم.

### Cursor چیست؟

Cursor ابزاری است که به کمک آن دستورات SQL را اجرا می‌کنیم و نتیجه‌ی بعضی از آن‌ها را دریافت می‌کنیم.

## ۸. مرحله چهارم: ساخت جدول

برای ساخت جدول از دستور SQL زیر استفاده می‌کنیم:

```sql
CREATE TABLE
```

جدول `students` شامل این ستون‌هاست:

- `id` → شناسه
- `name` → نام
- `age` → سن
- `score` → نمره

</div>

````python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER,
        score REAL
    )
""")

connection.commit()

print("The students table is ready.")
````

<div dir="rtl">

### چند بخش مهم این دستور

`PRIMARY KEY` یعنی این ستون شناسه‌ی یکتای رکوردهاست.

`AUTOINCREMENT` باعث می‌شود مقدار `id` به‌صورت خودکار افزایش پیدا کند:

```text
1
2
3
4
...
```

`IF NOT EXISTS` هم باعث می‌شود اگر جدول قبلاً وجود داشت، دوباره ساخته نشود.

</div>

<div dir="rtl">

## ۹. مرحله پنجم: اضافه کردن اطلاعات با `INSERT`

برای اضافه کردن رکورد از:

```sql
INSERT INTO
```

استفاده می‌کنیم.

در SQLite برای مقدارهایی که از Python می‌آیند از `?` استفاده می‌کنیم.

</div>

````python
sql = """
INSERT INTO students (name, age, score)
VALUES (?, ?, ?)
"""

values = ("علی", 16, 18.5)

cursor.execute(sql, values)

connection.commit()

print("Student added successfully.")
````

<div dir="rtl">

### چرا از `?` استفاده می‌کنیم؟

بهتر است مقدارهای Python را مستقیم داخل رشته‌ی SQL قرار ندهیم.

این روش را انجام ندهید:

```python
name = "Ali"
cursor.execute(f"INSERT INTO students (name) VALUES ('{name}')")
```

روش درست:

```python
cursor.execute(
    "INSERT INTO students (name) VALUES (?)",
    (name,)
)
```

این روش برای جلوگیری از مشکلاتی مانند **SQL Injection** هم اهمیت دارد.

</div>

<div dir="rtl">

## ۱۰. اضافه کردن چند رکورد

می‌توانیم چند دانش‌آموز را با چند اجرای `execute()` اضافه کنیم.

</div>

````python
students = [
    ("سارا", 17, 19),
    ("رضا", 16, 15.5),
    ("مریم", 17, 17),
    ("امیر", 16, 14)
]

for student in students:
    cursor.execute(
        "INSERT INTO students (name, age, score) VALUES (?, ?, ?)",
        student
    )

connection.commit()

print("Data added successfully.")
````

<div dir="rtl">

## ۱۱. `commit()` چیست؟

دستورهای:

```sql
INSERT
UPDATE
DELETE
```

اطلاعات دیتابیس را تغییر می‌دهند.

بعد از این تغییرات از:

```python
connection.commit()
```

استفاده می‌کنیم تا تغییرات ثبت شوند.

پس به‌صورت ساده:

```text
execute()
   ↓
تغییر
   ↓
commit()
   ↓
ثبت تغییر
```

</div>

## ۱۲. خواندن اطلاعات با `SELECT`

برای خواندن اطلاعات از:

```sql
SELECT
```

استفاده می‌کنیم.

</div>

````python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
````

<div dir="rtl">

## ۱۳. `fetchall()` چیست؟

بعد از اجرای `SELECT`، نتیجه را می‌توانیم با:

```python
cursor.fetchall()
```

دریافت کنیم.

`fetchall()` همه‌ی رکوردهای نتیجه را برمی‌گرداند.

</div>

<div dir="rtl">

## ۱۴. خواندن یک رکورد با `fetchone()`

`fetchone()` یکی از مهم‌ترین متدهای این جلسه است.

این متد فقط **یک رکورد** از نتیجه را برمی‌گرداند.

</div>

````python
cursor.execute("SELECT * FROM students")

row = cursor.fetchone()

print(row)
````

<div dir="rtl">

### تفاوت `fetchone()` و `fetchall()`

اگر نتیجه این باشد:

```text
(1, 'علی', 16, 18.5)
(2, 'سارا', 17, 19.0)
(3, 'رضا', 16, 15.5)
```

با:

```python
fetchone()
```

یک رکورد می‌گیریم.

با:

```python
fetchall()
```

تمام رکوردهای نتیجه را می‌گیریم.

</div>

## ۱۵. خواندن اطلاعات با شرط — `WHERE`

برای انتخاب رکوردهای موردنظر از:

```sql

```

استفاده می‌کنیم.

مثلاً دانش‌آموزانی که نمره‌شان بیشتر از ۱۷ است:

</div>

<div dir="rtl">

## ۱۶. ویرایش اطلاعات با `UPDATE`

برای تغییر اطلاعات از:

```sql
UPDATE
```

استفاده می‌کنیم.

مثلاً نمره‌ی رضا را تغییر می‌دهیم.

</div>

````python
cursor.execute(
    "UPDATE students SET score = ? WHERE name = ?",
    (18, "رضا")
)

connection.commit()

print("Data updated successfully.")
````

<div dir="rtl">

### هشدار مهم درباره‌ی `UPDATE`

همیشه به `WHERE` دقت کنید.

این دستور:

```sql

```

می‌تواند نمره‌ی **تمام رکوردها** را تغییر دهد.

اما:

```sql

SET score = 20
WHERE name = ?
```

فقط رکوردهای مطابق شرط را تغییر می‌دهد.

# ۲۱. پروژه‌ی جمع‌بندی — سیستم مدیریت دانش‌آموزان

در این پروژه تمام مفاهیم اصلی جلسه را کنار هم قرار می‌دهیم:

1. ساخت/باز کردن دیتابیس
2. ساخت جدول
3. اضافه کردن اطلاعات
4. خواندن اطلاعات
5. استفاده از `fetchone()`
6. ویرایش اطلاعات
7. حذف اطلاعات
8. بستن اتصال

</div>

````python
import sqlite3

# ۱. اتصال به دیتابیس
connection = sqlite3.connect("school_project.db")

# ۲. ساخت Cursor
cursor = connection.cursor()

# ۳. ساخت جدول
cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER,
        score REAL
    )
""")

# ۴. اضافه کردن چند دانش‌آموز
students = [
    ("علی", 16, 18.5),
    ("سارا", 17, 19),
    ("رضا", 16, 15.5)
]

cursor.executemany(
    "INSERT INTO students (name, age, score) VALUES (?, ?, ?)",
    students
)

connection.commit()

# ۵. نمایش همه‌ی دانش‌آموزان
cursor.execute("SELECT * FROM students")

print("All students:")
for student in cursor.fetchall():
    print(student)

# ۶. پیدا کردن یک دانش‌آموز با fetchone()
cursor.execute(
    "SELECT * FROM students WHERE name = ?",
    ("سارا",)
)

student = cursor.fetchone()

print("\nStudent found:")
print(student)

# ۷. ویرایش نمره
cursor.execute(
    "UPDATE students SET score = ? WHERE name = ?",
    (20, "سارا")
)

connection.commit()

# ۸. حذف یک دانش‌آموز
cursor.execute(
    "DELETE FROM students WHERE name = ?",
    ("رضا",)
)

connection.commit()

# ۹. نمایش نتیجه‌ی نهایی
print("\nFinal data:")
cursor.execute("SELECT * FROM students")

for student in cursor.fetchall():
    print(student)

# ۱۰. بستن اتصال
cursor.close()
connection.close()
````

## نکته‌ی اختیاری — شروع دوباره از صفر

اگر در طول تدریس خواستید دیتابیس تمرینی را کاملاً از نو بسازید، می‌توانید فایل `.db` را از پوشه‌ی پروژه حذف کنید و برنامه را دوباره اجرا کنید.

مثلاً:

```text

```

را حذف کنید؛ سپس `connect("school.db")` دوباره یک دیتابیس خالی می‌سازد.

> این روش برای تمرین مناسب است؛ در پروژه‌ی واقعی نباید بدون توجه به اطلاعات موجود، فایل دیتابیس را حذف کنید.
