##### در این جلسه با مفهوم **رشته (String)** و نحوه‌ی کار با آن در پایتون آشنا می‌شویم.

#### موضوعات این جلسه:

- ساخت رشته و رشته‌های چندخطی
- رشته‌ها به‌عنوان آرایه‌ای از کاراکترها
- حلقه زدن روی رشته
- طول رشته با `len()`
- بررسی وجود یک کاراکتر با `in` و `not in`

- تمرین عملی

## ساخت رشته

##### رشته‌ها در پایتون می‌توانند هم با **کوتیشن تکی (`'`)** و هم با **کوتیشن دوتایی (`"`)** نوشته شوند.

##### هر دو روش کاملاً معتبر هستند و هیچ تفاوتی با هم ندارند.

</div>


```python
print("Hello")
print('Hello')
```

<div dir="rtl">

## کوتیشن داخل کوتیشن

##### می‌توان داخل یک رشته از کوتیشن استفاده کرد، به شرطی که با کوتیشن بیرونی **یکسان نباشد.**

</div>


```python
print("It's alright")
print("He is called 'Johnny'")
print('He is called "Johnny"')
```

<div dir="rtl">

## اختصاص رشته به متغیر

##### مثل هر مقدار دیگری، رشته هم با علامت `=` به یک متغیر اختصاص داده می‌شود.

</div>


```python
a = "Hello"
print(a)
```

<div dir="rtl">

## رشته‌های چندخطی (Multiline Strings)

##### با استفاده از **سه کوتیشن** (تکی یا دوتایی) می‌توان یک رشته را در **چند خط** نوشت.

##### در نتیجه، خط‌های جدید دقیقاً در همان جایی که در کد نوشته شده‌اند، ظاهر می‌شوند.

</div>


```python
a = """سطر اول متن
سطر دوم متن
سطر سوم متن"""

print(a)
```

<div dir="rtl">

## 🧠 نکته — کوتیشن سه‌تایی تکی هم کار می‌کند

##### به‌جای سه کوتیشن دوتایی (`"""`) می‌توان از سه کوتیشن تکی (`'''`) هم استفاده کرد.

</div>


```python
a = '''سطر اول متن
سطر دوم متن'''

print(a)
```

<div dir="rtl">

## رشته‌ها آرایه‌ای از کاراکترها هستند

##### در پایتون، مثل خیلی از زبان‌های دیگر، **رشته‌ها آرایه‌ای از کاراکترها** هستند.

##### با این حال، پایتون یک **نوع داده جداگانه برای "کاراکتر تکی"** ندارد؛ هر کاراکتر تکی هم خودش یک رشته با طول ۱ است.

##### با استفاده از **کروشه `[]`** می‌توان به کاراکترهای یک رشته دسترسی پیدا کرد.

</div>


```python
a = "Hello, World!"

# گرفتن کاراکتر در جایگاه ۱ (اولین کاراکتر جایگاه ۰ دارد)
print(a[1])
```

<div dir="rtl">

## حلقه زدن روی رشته

##### چون رشته‌ها آرایه هستند، می‌توانیم با حلقه‌ی `for` روی کاراکترهای آن حلقه بزنیم.

##### توضیحات کامل حلقه‌ها را در جلسات آینده می‌بینیم؛ اینجا فقط با یک کاربرد ساده از آن آشنا می‌شویم.

</div>


```python
for x in "banana":
    print(x)
```

<div dir="rtl">

## طول رشته — `len()`

##### برای گرفتن **طول یک رشته** (تعداد کاراکترهای آن) از تابع `len()` استفاده می‌کنیم.

</div>


```python
a = "Hello, World!"
print(len(a))
```

<div dir="rtl">

## بررسی وجود یک بخش در رشته — `in`

##### برای بررسی اینکه آیا یک **عبارت یا کاراکتر خاص** در یک رشته وجود دارد یا نه، از کلمه‌کلیدی `in` استفاده می‌کنیم.

</div>


```python
txt = "The best things in life are free!"
print("free" in txt)
```

<div dir="rtl">

##### همچنین می‌توان از `in` داخل یک دستور `if` هم استفاده کرد.

</div>


```python
txt = "The best things in life are free!"

if "free" in txt:
    print("بله، 'free' در متن وجود دارد.")
```

<div dir="rtl">

## بررسی عدم وجود — `not in`

##### برای بررسی اینکه یک عبارت **در رشته وجود ندارد**، از `not in` استفاده می‌کنیم.

</div>


```python
txt = "The best things in life are free!"

print("expensive" not in txt)

if "expensive" not in txt:
    print("نه، 'expensive' در متن وجود ندارد.")
```

## برش رشته (Slicing)

##### با استفاده از **Slicing** می‌توانیم بخشی از یک رشته را با مشخص کردن **جایگاه شروع و پایان** استخراج کنیم.

##### فرمت کلی: `رشته[شروع:پایان]`

##### نکته: جایگاه **پایان جزو نتیجه نیست.**

</div>


```python
b = "Hello, World!"

# گرفتن کاراکترهای جایگاه ۲ تا ۵ (جایگاه ۵ جزو نتیجه نیست)
print(b[2:5])
```

<div dir="rtl">

## برش از ابتدا

##### اگر **جایگاه شروع را ننویسیم**، برش از اولین کاراکتر شروع می‌شود.

</div>


```python
b = "Hello, World!"
print(b[:5])
```

<div dir="rtl">

## برش تا انتها

##### اگر **جایگاه پایان را ننویسیم**، برش تا آخرین کاراکتر ادامه پیدا می‌کند.

</div>


```python
b = "Hello, World!"
print(b[2:])
```

<div dir="rtl">

## اندیس‌گذاری منفی (Negative Indexing)

##### با استفاده از **اندیس‌های منفی** می‌توانیم برش را از **انتهای رشته** شروع کنیم.

##### اندیس `-1` به آخرین کاراکتر اشاره دارد.

</div>


```python
b = "Hello, World!"

# از کاراکتر "o" در "World!" (جایگاه -5)
# تا قبل از کاراکتر "d" در "World!" (جایگاه -2)
print(b[-5:-2])
```
