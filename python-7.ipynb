## متدهای رشته چیست؟

##### پایتون مجموعه‌ای از **متدهای آماده** برای کار روی رشته‌ها دارد.

##### نکته مهم: تمام متدهای رشته یک **مقدار جدید** برمی‌گردانند و **رشته‌ی اصلی را تغییر نمی‌دهند** (رشته‌ها Immutable هستند).

</div>


<div dir="rtl">

## حروف بزرگ — `upper()`

##### متد `upper()` رشته را به **حروف بزرگ** تبدیل می‌کند.

</div>


```python
a = "Hello, World!"
print(a.upper())
```

<div dir="rtl">

## حروف کوچک — `lower()`

##### متد `lower()` رشته را به **حروف کوچک** تبدیل می‌کند.

</div>


```python
a = "Hello, World!"
print(a.lower())
```

<div dir="rtl">

## حذف فاصله‌های اضافه — `strip()`

##### گاهی رشته‌ها در ابتدا یا انتها **فاصله‌ی اضافه (Whitespace)** دارند.

##### متد `strip()` این فاصله‌های اضافه را از ابتدا و انتهای رشته حذف می‌کند.

</div>


```python
a = "   Hello, World!   "
print(a.strip())  # نتیجه: "Hello, World!"
```

<div dir="rtl">

## جایگزینی — `replace()`

##### متد `replace()` یک بخش از رشته را با بخش دیگری **جایگزین** می‌کند.

</div>


```python
a = "Hello, World!"
print(a.replace("H", "J"))
```

<div dir="rtl">

## تقسیم رشته — `split()`

##### متد `split()` رشته را بر اساس یک **جداکننده (Separator)** تقسیم می‌کند و یک **لیست** برمی‌گرداند.

##### توضیحات کامل لیست‌ها را در جلسات بعدی می‌بینیم.

</div>


```python
a = "Hello, World!"
print(a.split(","))  # نتیجه: ['Hello', ' World!']
```

<div dir="rtl">

## چند متد پرکاربرد دیگر

##### علاوه‌بر متدهای بالا، این متدها هم خیلی زیاد استفاده می‌شوند:

- `count()` ← تعداد تکرار یک بخش در رشته را می‌شمارد
- `find()` ← جایگاه اولین وقوع یک بخش را پیدا می‌کند (اگر نبود، عدد `1-` برمی‌گرداند)
- `startswith()` / `endswith()` ← بررسی می‌کند رشته با یک بخش خاص شروع یا تمام می‌شود یا نه

</div>


```python
a = "Hello, World!"

print(a.count("l"))              # تعداد تکرار "l"
print(a.find("World"))           # جایگاه شروع "World"
print(a.startswith("Hello"))     # آیا با "Hello" شروع می‌شود؟
print(a.endswith("!"))           # آیا با "!" تمام می‌شود؟
```

<div dir="rtl">

##### 📌 لیست کامل متدهای رشته خیلی بیشتر از این‌هاست. در طول دوره، هر جا لازم شد، متد جدیدی رو معرفی می‌کنیم.

</div>

## الحاق رشته‌ها (Concatenation)

##### برای **ترکیب یا الحاق** دو رشته می‌توانیم از عملگر `+` استفاده کنیم.

</div>


```python
a = "Hello"
b = "World"
c = a + b
print(c)
```

<div dir="rtl">

##### برای اضافه کردن **فاصله** بین دو رشته، می‌توانیم یک `" "` هم به آن‌ها اضافه کنیم.

</div>


```python
a = "Hello"
b = "World"
c = a + " " + b
print(c)
```

## قالب‌بندی رشته با F-String

##### همان‌طور که در جلسه‌ی Casting دیدیم، نمی‌توانیم مستقیماً رشته و عدد را با `+` ترکیب کنیم.

</div>


```python
age = 36

# این خطا می‌دهد!
# txt = "اسم من جان است، " + age + " سال دارم."
# print(txt)
```

<div dir="rtl">

##### راه‌حل بهتر و مدرن‌تر برای این کار، استفاده از **F-String** است.

##### کافیست حرف `f` را قبل از رشته بنویسیم و متغیرها را داخل آکولاد `{}` قرار دهیم.

</div>


```python
age = 36
txt = f"اسم من جان است، {age} سال دارم."
print(txt)
```

<div dir="rtl">

## Placeholder و Modifier

##### داخل آکولاد `{}` می‌توان **متغیر، عملیات، یا حتی تابع** نوشت.

</div>


```python
price = 59
txt = f"قیمت این کالا {price} دلار است."
print(txt)
```

<div dir="rtl">

##### همچنین می‌توان با یک **Modifier** فرمت مقدار را هم مشخص کرد.

##### مثلاً `.2f` یعنی عدد اعشاری با دقیقاً **۲ رقم اعشار**.

</div>


```python
price = 59
txt = f"قیمت این کالا {price:.2f} دلار است."
print(txt)
```

<div dir="rtl">

##### داخل `{}` حتی می‌توان یک **عملیات ریاضی** هم نوشت.

</div>


```python
txt = f"قیمت کل این کالاها {20 * 59} دلار است."
print(txt)
```

## کاراکترهای فرار (Escape Characters)

##### گاهی می‌خواهیم داخل یک رشته از کاراکترهایی استفاده کنیم که **جزو نحو (Syntax) پایتون** هستند.

##### مثلاً کوتیشن دوتایی داخل یک رشته که با کوتیشن دوتایی شروع شده.

##### برای این کار از **کاراکتر فرار (`\`)** استفاده می‌کنیم.

</div>


```python
# بدون کاراکتر فرار خطا می‌دهد:
# txt = "ما به اصطلاح "وایکینگ" هستیم."

# با کاراکتر فرار درست کار می‌کند:
txt = "ما به اصطلاح \"وایکینگ\" هستیم."
print(txt)
```

<div dir="rtl">

## سایر کاراکترهای فرار پرکاربرد

| کد | نتیجه |
|----|-------|
| `\'` | کوتیشن تکی |
| `\\` | بک‌اسلش |
| `\n` | خط جدید (New Line) |
| `\t` | تب (Tab) |

</div>


```python
print("سطر اول\nسطر دوم")
print("نام:\tعلی")
```

## 💻 تمرین کلاسی

##### کد زیر را کامل کنید:

</div>


```python
name = "  ali  "

# ۱. فاصله‌های اضافه‌ی name را حذف کنید
clean_name = name.___()

# ۲. clean_name را با متد مناسب به حروف بزرگ تبدیل و چاپ کنید
print(clean_name.___())

# ۳. با استفاده از F-String یک جمله بسازید: "نام کاربر: <clean_name>"
print(___"نام کاربر: {clean_name}")

# ۴. دو رشته‌ی "Python" و "Course" را با یک فاصله به هم الحاق کنید
result = "Python" ___ " " ___ "Course"
print(result)
```
