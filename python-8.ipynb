## مقادیر Boolean

##### در برنامه‌نویسی، اغلب لازم است بدانیم یک عبارت **درست (True)** است یا **نادرست (False)**.

##### هر عبارتی در پایتون در نهایت به یکی از این دو مقدار ارزیابی می‌شود.

</div>


```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

<div dir="rtl">

##### همچنین، وقتی شرط یک دستور `if` را اجرا می‌کنیم، پایتون همان `True` یا `False` را برمی‌گرداند.

</div>


```python
a = 200
b = 33

if b > a:
    print("b بزرگتر از a است")
else:
    print("b بزرگتر از a نیست")
```

<div dir="rtl">

## ارزیابی مقادیر با `bool()`

##### تابع `bool()` هر مقداری را می‌گیرد و نتیجه‌ی `True` یا `False` آن را برمی‌گرداند.

</div>


```python
print(bool("Hello"))
print(bool(15))
```

<div dir="rtl">

## بیشتر مقادیر True هستند

##### تقریباً هر مقداری که **محتوایی** داشته باشد، `True` ارزیابی می‌شود:

- هر رشته‌ای `True` است، به جز رشته‌ی خالی (`""`)
- هر عددی `True` است، به جز `0`
- هر لیست، تاپل، مجموعه یا دیکشنری‌ای `True` است، به جز موارد خالی

</div>


```python
print(bool("abc"))
print(bool(123))
print(bool(["apple", "cherry", "banana"]))
```

<div dir="rtl">

## چند مقدار False هستند

##### مقادیر خیلی کمی هستند که `False` ارزیابی می‌شوند: مقادیر **خالی** مثل `()`, `[]`, `{}`, `""`, عدد `0`، و مقدار `None`.

</div>


```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool([]))
print(bool({}))
```

<div dir="rtl">

## توابع هم می‌توانند Boolean برگردانند

##### تابع `isinstance()` یکی از توابع پرکاربرد پایتون است که بررسی می‌کند یک مقدار از یک نوع داده‌ی خاص هست یا نه، و `True`/`False` برمی‌گرداند.

</div>


```python
x = 200
print(isinstance(x, int))
```

## عملگرها (Operators) چیستند؟

##### عملگرها برای انجام **عملیات روی متغیرها و مقادیر** استفاده می‌شوند.

##### پایتون عملگرها را به این دسته‌ها تقسیم می‌کند:

- عملگرهای ریاضی (Arithmetic)
- عملگرهای انتساب (Assignment)
- عملگرهای مقایسه‌ای (Comparison)
- عملگرهای منطقی (Logical)
- عملگرهای هویت (Identity)
- عملگرهای عضویت (Membership)
- عملگرهای بیتی (Bitwise)

##### در این جلسه مهم‌ترین این دسته‌ها را بررسی می‌کنیم.

</div>


<div dir="rtl">

## عملگرهای ریاضی (Arithmetic)

| عملگر | نام | مثال |
|---|---|---|
| `+` | جمع | `x + y` |
| `-` | تفریق | `x - y` |
| `*` | ضرب | `x * y` |
| `/` | تقسیم | `x / y` |
| `%` | باقی‌مانده (Modulus) | `x % y` |
| `**` | توان | `x ** y` |
| `//` | تقسیم صحیح (Floor Division) | `x // y` |

</div>


```python
x = 15
y = 4

print(x + y)   # جمع
print(x - y)   # تفریق
print(x * y)   # ضرب
print(x / y)   # تقسیم
print(x % y)   # باقی‌مانده
print(x ** y)  # توان
print(x // y)  # تقسیم صحیح
```

<div dir="rtl">
<hr>

## عملگرهای انتساب (Assignment)

##### این عملگرها برای **مقداردهی** به متغیرها استفاده می‌شوند و می‌توانند همزمان یک عملیات هم انجام دهند.

| عملگر | معادل | مثال |
|---|---|---|
| `=` | — | `x = 5` |
| `+=` | `x = x + y` | `x += 3` |
| `-=` | `x = x - y` | `x -= 3` |
| `*=` | `x = x * y` | `x *= 3` |
| `/=` | `x = x / y` | `x /= 3` |
| `//=` | `x = x // y` | `x //= 3` |
| `**=` | `x = x ** y` | `x **= 3` |

</div>


```python
x = 10
print(x)

x += 5
print(x)  # معادل x = x + 5

x -= 2
print(x)  # معادل x = x - 2

x *= 3
print(x)  # معادل x = x * 3
```

<div dir="rtl">
<hr>

## عملگرهای مقایسه‌ای (Comparison)

##### این عملگرها دو مقدار را با هم مقایسه می‌کنند و همیشه **یک مقدار Boolean** برمی‌گردانند.

| عملگر | نام | مثال |
|---|---|---|
| `==` | برابر | `x == y` |
| `!=` | نابرابر | `x != y` |
| `>` | بزرگتر از | `x > y` |
| `<` | کوچکتر از | `x < y` |
| `>=` | بزرگتر یا مساوی | `x >= y` |
| `<=` | کوچکتر یا مساوی | `x <= y` |

</div>


```python
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

<div dir="rtl">
<hr>

## عملگرهای منطقی (Logical)

##### این عملگرها برای **ترکیب چند شرط** استفاده می‌شوند.

| عملگر | توضیح | مثال |
|---|---|---|
| `and` | هر دو شرط باید درست باشند | `x < 5 and x < 10` |
| `or` | حداقل یکی از شرط‌ها باید درست باشد | `x < 5 or x < 4` |
| `not` | نتیجه‌ی شرط را معکوس می‌کند | `not(x < 5 and x < 10)` |

</div>


```python
x = 5

print(x < 10 and x > 2)   # هر دو درست‌اند -> True
print(x < 4 or x > 2)     # یکی درست است -> True
print(not(x < 10))        # معکوس True -> False
```

<div dir="rtl">
<hr>

## عملگرهای هویت (Identity)

##### عملگرهای `is` و `is not` بررسی می‌کنند آیا دو متغیر به **یک شیء یکسان در حافظه** اشاره می‌کنند یا نه (نه اینکه مقدارشان برابر است).

</div>


```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x == y)   # مقدارشان برابر است -> True
print(x is y)   # ولی دو شیء متفاوت‌اند -> False
print(x is z)   # z همان x است -> True
```

<div dir="rtl">
<hr>

## عملگرهای عضویت (Membership)

##### با `in` و `not in` بررسی می‌کنیم آیا یک مقدار در یک دنباله (رشته، لیست و...) وجود دارد یا نه.

##### در جلسه‌ی رشته‌ها با `in` برای رشته‌ها آشنا شدیم؛ همین عملگر روی لیست، تاپل و... هم کار می‌کند.

</div>


```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
print("melon" not in fruits)
```

## 💻 تمرین کلاسی

##### کد زیر را کامل کنید:

</div>


```python
age = 20
has_ticket = True

# ۱. بررسی کنید آیا age بزرگتر یا مساوی ۱۸ است
is_adult = age ___ 18
print(is_adult)

# ۲. با عملگر منطقی and بررسی کنید آیا فرد بزرگسال است و بلیط هم دارد
can_enter = is_adult ___ has_ticket
print(can_enter)

# ۳. مقدار age را با استفاده از عملگر انتساب += به‌اندازه ۱ افزایش دهید
age ___ 1
print(age)

# ۴. بررسی کنید آیا عدد age بر ۲ بخش‌پذیر است (باقی‌مانده صفر است)
print(age ___ 2 == 0)
```

## جمع‌بندی

##### در این جلسه یاد گرفتیم:

- Booleanها فقط دو مقدار دارند: `True` و `False`
- تابع `bool()` هر مقداری را به `True` یا `False` تبدیل می‌کند؛ مقادیر خالی و `0` معمولاً `False` هستند
- عملگرهای ریاضی برای محاسبات پایه استفاده می‌شوند (`+ - * / % ** //`)
- عملگرهای انتساب مقداردهی و عملیات را همزمان انجام می‌دهند (`+= -= *=` و...)
- عملگرهای مقایسه‌ای همیشه Boolean برمی‌گردانند (`== != > < >= <=`)
- عملگرهای منطقی چند شرط را ترکیب می‌کنند (`and or not`)
- `is` هویت (یکسان بودن شیء در حافظه) را بررسی می‌کند، نه فقط برابری مقدار
- `in` و `not in` عضویت یک مقدار در یک دنباله را بررسی می‌کنند
