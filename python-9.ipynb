# جلسه ۹ — لیست‌ها (Lists) در پایتون — بخش ۱

##### در این جلسه با **لیست** به‌عنوان اولین ساختمان داده‌ی پایتون آشنا می‌شویم.

#### موضوعات این جلسه:

- لیست چیست و چگونه ساخته می‌شود؟
- ویژگی‌های لیست (ترتیب‌دار، قابل تغییر، مجاز به تکرار)
- طول لیست با `len()`
- دسترسی به آیتم‌ها با ایندکس
- اندیس‌گذاری منفی
- برش لیست (Slicing)
- بررسی وجود یک آیتم با `in`
- تمرین عملی

</div>


<div dir="rtl">

## لیست چیست؟

##### **لیست (List)** برای ذخیره‌ی **چندین مقدار** در یک متغیر استفاده می‌شود.

##### لیست‌ها یکی از ۴ نوع داده‌ی مجموعه‌ای (Collection) در پایتون هستند؛ سه مورد دیگر Tuple، Set و Dictionary هستند که در جلسات بعد می‌بینیم.

##### لیست‌ها با **کروشه `[]`** ساخته می‌شوند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
print(my_list)
```

<div dir="rtl">

## ویژگی‌های لیست

##### آیتم‌های لیست **ایندکس‌دار (Indexed)** هستند؛ اولین آیتم ایندکس `0` دارد.

##### لیست‌ها سه ویژگی مهم دارند:

- **ترتیب‌دار (Ordered):** ترتیب آیتم‌ها ثابت است و تغییر نمی‌کند
- **قابل تغییر (Changeable):** می‌توانیم آیتم‌ها را تغییر دهیم، اضافه کنیم یا حذف کنیم
- **مجاز به تکرار (Allow Duplicates):** چون ایندکس‌دار هستند، می‌توانند مقدار تکراری هم داشته باشند

</div>


```python
my_list = ["apple", "banana", "cherry", "apple", "cherry"]
print(my_list)
```

<div dir="rtl">

## طول لیست — `len()`

##### برای گرفتن **تعداد آیتم‌های** یک لیست از تابع `len()` استفاده می‌کنیم — دقیقاً مثل رشته‌ها.

</div>


```python
thislist = ["apple", "banana", "cherry"]
print(len(thislist))
```

<div dir="rtl">

## لیست می‌تواند نوع داده‌های مختلف داشته باشد

##### آیتم‌های یک لیست می‌توانند از **هر نوع داده‌ای** باشند.

</div>


```python
list1 = ["apple", "banana", "cherry"]
list2 = [1, 5, 7, 9, 3]
list3 = [True, False, False]

print(list1)
print(list2)
print(list3)
```

<div dir="rtl">

##### یک لیست حتی می‌تواند **ترکیبی از چند نوع داده** باشد.

</div>


```python
list1 = ["abc", 34, True, 40, "male"]
print(list1)
```

<div dir="rtl">

## نوع داده‌ی لیست — `type()`

##### از نگاه پایتون، لیست‌ها یک شیء با نوع داده‌ی `'list'` هستند.

</div>


```python
my_list = ["apple", "banana", "cherry"]
print(type(my_list))
```

<div dir="rtl">

## ساخت لیست با تابع سازنده — `list()`

##### مثل رشته‌ها و اعداد، لیست هم یک **تابع سازنده** دارد.

##### توجه: در این روش از **دو پرانتز** استفاده می‌شود.

</div>


```python
my_list = list(("apple", "banana", "cherry"))
print(my_list)
```

<div dir="rtl">
<hr>

## دسترسی به آیتم‌ها (Access Items)

##### چون آیتم‌های لیست ایندکس‌دار هستند، می‌توانیم با نوشتن **شماره ایندکس** به آن‌ها دسترسی پیدا کنیم.

</div>


```python
my_list = ["apple", "banana", "cherry"]
print(my_list[1])
```

<div dir="rtl">

## اندیس‌گذاری منفی (Negative Indexing)

##### `-1` به **آخرین آیتم** اشاره می‌کند، `-2` به آیتم ماقبل آخر و به همین ترتیب.

</div>


```python
my_list = ["apple", "banana", "cherry"]
print(my_list[-1])
```

<div dir="rtl">
<hr>

## بازه‌ای از ایندکس‌ها (Range of Indexes)

##### می‌توانیم با مشخص کردن ایندکس شروع و پایان، بخشی از لیست را برگردانیم.

##### خروجی همیشه یک **لیست جدید** خواهد بود.

</div>


```python
my_list = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]

print(my_list[2:5])
```

<div dir="rtl">

##### نکته: جست‌وجو از ایندکس ۲ **شروع** می‌شود و در ایندکس ۵ (که خودش جزو نتیجه نیست) **تمام** می‌شود.

</div>


```python
my_list = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]

print(my_list[:4])

print(my_list[2:])
```

<div dir="rtl">

## بازه‌ی ایندکس‌های منفی

##### می‌توانیم برای برش، از ایندکس‌های منفی هم استفاده کنیم.

</div>


```python
my_list = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]

print(my_list[-4:-1])
```

<div dir="rtl">
<hr>

## بررسی وجود یک آیتم — `in`

##### برای بررسی اینکه آیا یک مقدار در لیست وجود دارد، از کلمه‌کلیدی `in` استفاده می‌کنیم — دقیقاً مثل رشته‌ها.

</div>


```python
my_list = ["apple", "banana", "cherry"]

if "apple" in my_list:
    print("yes, apple is on the list")
```

## جمع‌بندی

##### در این جلسه یاد گرفتیم:

- لیست‌ها با `[]` ساخته می‌شوند و برای ذخیره‌ی چند مقدار در یک متغیر استفاده می‌شوند
- لیست‌ها ترتیب‌دار، قابل تغییر و مجاز به تکرار هستند
